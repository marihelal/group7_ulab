## Line Extraction Tool

#### To use this tool, download the lines of the molecules of interest from CASSIS as a txt file. Simply paste the name of the file in the file_name variable below and run the code. The data will be automatically downloaded and ready to be pasted into excel!

```python
import numpy as np
import pandas as pd
```

```python
file_name='file_name.txt'
df = pd.read_csv(file_name, sep="\t", header=1)

df=df.loc[:, ['Transition', 'Species_id', 'Frequency(MHz)', 'Eup(K)', 'Aij(s-1)']] 
df['Frequency(MHz)']=df['Frequency(MHz)']*10**6
df = df.rename(columns={
    'Transition': 'Name',
    'Species_id': 'Tag',
    'Frequency(MHz)': 'v (Hz)',
    'Eup(K)': 'Eup (K)',
    'Aij(s-1)': 'Aij (s-1)'})
df=df.reindex(columns=['Name','Tag','Eup (K)', 'Aij (s-1)','v (Hz)'] )

df.to_clipboard(index=False, header=False)
df
```
